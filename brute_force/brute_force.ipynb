```python
import numpy as np
import pandas as pd

import brute_force
```

```python
users_df = pd.read_csv("../data/v-small/amazon_embeddings_users.csv")
items_df = pd.read_csv("../data/v-small/amazon_embeddings.csv")
```

```python
# Load data (reuse the same dataframes)
u_mat, i_mat, i_bias = brute_force.load_data(users_df, items_df)

# Define your ads
configs = []
for i in range(2):  # Creating 2 ads for this tiny demo
    configs.append({
        'id': i,
        'min_users': 1,
        'max_users': 5
    })
```

```python
# Call exact solver
best_u_exact, best_i_exact, score_exact = brute_force.brute_force_optimize(
    u_mat, i_mat, i_bias, configs, items_per_ad=2
)

print(best_u_exact)
print(best_i_exact)
print(score_exact)
```
